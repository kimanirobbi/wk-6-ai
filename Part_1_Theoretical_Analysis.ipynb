# Essay Questions.

**Question 1**

***1: Explain how Edge AI reduces latency and enhances privacy compared to cloud-based AI. Provide a real-world example.***

**Answers**

- Latency: In cloud AI, data travels from your device to a remote server and back, causing delays. Edge AI processes data locally on the device (e.g., the drone's own computer), eliminating this round trip and making decisions instantly.

- Privacy: With Edge AI, sensitive data (e.g., video feeds, personal conversations) never leaves your device. In cloud AI, this data is transmitted and stored on external servers, creating a privacy risk.

- Real world examples (traffic management systems.):A traffic camera using cloud AI would have to stream live video to the cloud to identify objects. This could be slow (high latency) and if the connection drops, the management system becomes "blind." Smart traffic lights with edge AI cameras analyze real-time traffic flow and pedestrian movement to dynamically adjust signal timings, reducing congestion and improving safety.


**Question 2**

***2: Compare Quantum AI and classical AI in solving optimization problems. What industries could benefit most from Quantum AI?***

**Answers**

- Classical AI (e.g., on standard computers): Excels at many tasks but struggles with certain types of complex optimization problems (e.g., finding the absolute best solution among billions of possibilities). It often has to settle for a "good enough" solution.

- Quantum AI: Leverages quantum mechanics (superposition, entanglement) to evaluate multiple possibilities simultaneously. This makes it potentially revolutionary for problems where the number of variables is enormous.

**Industries that could benefit:**

**1.** Pharmaceuticals & Personalized Medicine: Simulating molecular interactions for drug discovery.

**2.** Finance: Optimizing complex investment portfolios and managing risk.

**3.** Logistics & Supply Chain: Solving the "traveling salesman" problem on a global scale for optimal routing.

**4.** Materials Science: Designing new materials with specific properties.
